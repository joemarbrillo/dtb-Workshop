Simulate change in the source system: some customers move tier.

```python
import duckdb
```

```python
con = duckdb.connect("shopflow.duckdb")
```

```python
con.execute("""
            update raw.customers
            set tier = 'gold', updated_at = current_timestamp::timestamp
            where customer_id in (1,2,3)
            """)
```

```python
print(con.sql("select customer_id, tier, updated_at from raw.customers where customer_id <= 3"))
```

```python
con.close()
```
