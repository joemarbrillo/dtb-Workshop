Export the monthly report revenue mart to a csv for the business.

```python
from pathlib import Path
import duckdb
```

```python
Path("reports").mkdir(exist_ok=True)
con = duckdb.connect("shopflow.duckdb", read_only=True)
```

```python
con.execute("""
            copy (select * from marts.mart_monthly_revenue)
            to 'reports/monthly_revenue.csv' (header, delimiter ',')
            """)
print(con.sql("""
              select region,
              sum(orders) as order,
              round(sum(revenue), 2) as revenue
              from marts.mart_monthly_revenue group by 1 order by 3 desc
              """))
con.close()
```
